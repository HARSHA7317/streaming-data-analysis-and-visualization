**Visualization**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv('/content/drive/MyDrive/energydata_complete.csv')
data['date'] = pd.to_datetime(data['date'])
data['hour'] = data['date'].dt.hour

hourly_energy = data.groupby('hour')['Appliances'].mean()

plt.figure(figsize = (14,5))

plt.bar(hourly_energy.index,hourly_energy.values)

plt.xlabel("Hour")
plt.ylabel("Energy Consumption")
plt.title("Energy Consumption by Hour")

plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv('/content/drive/MyDrive/energydata_complete.csv')
stream_data = data[['Appliances','date']].copy()

stream_data['date'] = pd.to_datetime(stream_data['date'])
stream_data['moving_average'] = stream_data['Appliances'].rolling(window = 10).mean()

plt.figure(figsize = (14,5))

plt.plot(stream_data['date'],stream_data['Appliances'],label = 'Energy Consumption')
plt.plot(stream_data['date'],stream_data['moving_average'],label = 'Moving Average')

plt.xlabel("Time")
plt.ylabel("Energy Consumption (Wh)")
plt.title("Household Appliance Energy Consumption Over Time")

plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv('/content/drive/MyDrive/energydata_complete.csv')
stream_data = data[['Appliances','date']].copy()

stream_data['date'] = pd.to_datetime(stream_data['date'])
mean_energy = np.mean(stream_data['Appliances'])
std_energy = np.std(stream_data['Appliances'])

treshold = mean_energy + 2 * std_energy

stream_data['anamolies'] = (stream_data['Appliances'] > treshold)
anomalies = stream_data[stream_data['anamolies']]

plt.figure(figsize = (14,5))

plt.plot(stream_data['date'],stream_data['Appliances'],label = 'Energy Consumption')
plt.plot(anomalies['date'],anomalies['Appliances'],label = 'Potential Anamoly')

plt.xlabel("Time")
plt.ylabel("Energy Consumption (Wh)")
plt.title("Energy Consumption with Potential Anomalies")

plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('/content/drive/MyDrive/energydata_complete.csv')

data['date'] = pd.to_datetime(data['date'])
daily_data = data.set_index('date')['Appliances'].resample('D').mean().reset_index()

subset = daily_data.head(30)

plt.figure(figsize=(14, 5))

plt.bar(subset['date'], subset['Appliances'], color='tab:blue', width=0.6, label='Daily Avg Appliances Energy')

plt.xlabel("Date", fontsize=12)
plt.ylabel("Energy Consumption (Wh)", fontsize=12)
plt.title("Daily Average Household Appliance Energy Consumption", fontsize=14, fontweight='bold')

plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('/content/drive/MyDrive/energydata_complete.csv')

bins = [0, 50, 100, 200, 500, float('inf')]
labels = ['Low (0-50 Wh)', 'Moderate (50-100 Wh)', 'High (100-200 Wh)', 'Very High (200-500 Wh)', 'Extreme (>500 Wh)']
data['Usage_Category'] = pd.cut(data['Appliances'], bins=bins, labels=labels)

category_counts = data['Usage_Category'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=plt.cm.Paired.colors
)

plt.title("Proportion of Appliance Energy Consumption Ranges", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
